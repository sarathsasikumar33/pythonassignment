```python
import pandas as pd
```

```python
import numpy as np
```

```python
import matplotlib.pyplot as plt
```

```python
import seaborn as sns
```

```python
df = sns.load_dataset("taxis")
```

```python
df.head()
```

```python
df.info()
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
df.dropna(subset=["pickup","fare","distance"],inplace=True)
```

```python
df.info()
```

```python
line_df=df.dropna(subset=["pickup","fare"]).sort_values("pickup")
plt.figure(figsize=(12,6))
plt.plot(line_df["pickup"],line_df["fare"])
plt.title("Fare Over Time")
plt.xlabel("Pickup Timestamp")
plt.ylabel("Fare")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
borough_fare=df.groupby("pickup_borough",observed=False)["fare"].sum()
borough_fare.plot(kind="bar",figsize=(10,6))
plt.title("Total Fare by Pickup Borough")
plt.xlabel("Pickup Borough")
plt.ylabel("Total Fare")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
payment_counts=df["payment"].value_counts()
plt.figure(figsize=(8,8))
plt.pie(payment_counts,labels=payment_counts.index,autopct="%1.1f%%",startangle=90)
plt.title("Distribution of Trips by Payment Method")
plt.axis("equal")
plt.show()
```

```python
plt.figure(figsize=(10,6))
plt.hist(df["distance"].dropna(),bins=20,edgecolor="black")
plt.title("Distribution of Taxi Trip Distances")
plt.xlabel("Distance")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.boxplot(data=df,x="pickup_borough",y="tip")
plt.title("Distribution of Trips by Pickup Borough")
plt.xlabel("Pickup Borough")
plt.ylabel("Tip Amount")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.countplot(data=df,x="pickup_borough")
plt.title("Number of Trips by Pickup Borough")
plt.xlabel("Pickup Borough")
plt.ylabel("Count of Trips")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot(data=df,x="distance",y="fare",hue="pickup_borough")
plt.title("Relationship b/w Distance and Fare")
plt.xlabel("Distance")
plt.ylabel("Fare")
plt.tight_layout()
plt.show()
```

```python
columns=["distance","fare","tip","tolls","total"]
correlation_matrix=df[columns].corr()
plt.figure(figsize=(10,6))
sns.heatmap(correlation_matrix,annot=True,cmap="coolwarm",fmt=".2f")
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```

```python
pair_df=df[["distance","fare","tip","total","pickup_zone"]].dropna()
sns.pairplot(pair_df,vars=["distance","fare","tip","total"],
                 hue="pickup_zone",diag_kind="hist")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.violinplot(data=df,x="payment",y="fare")
plt.title("Distribution of Fares by Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Fare")
plt.tight_layout()
plt.show()
```
