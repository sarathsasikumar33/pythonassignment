```python
import numpy as np
```

```python
temperatures_w1=np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
```

```python
print(temperatures_w1)
```

```python
print("Shape:",temperatures_w1.shape)
print("Data Type:",temperatures_w1.dtype)
print("Number of Elements:",temperatures_w1.size)
```

```python
fahrenheit=(temperatures_w1*9/5)+32
print(fahrenheit)
```

```python
maximum=np.max(temperatures_w1)
minimum=np.min(temperatures_w1)
mean=np.mean(temperatures_w1)

print("Maximum Temperature:",maximum)
print("Minimum Temperature:",minimum)
print("Mean Temperature:",mean)
```

```python
first_three_days=temperatures_w1[:3]
weekend=temperatures_w1[-2:]
middle_three_days=temperatures_w1[2:5]

print("First three days:",first_three_days)
print("Weekend:",weekend)
print("Middle three days:",middle_three_days)
```

```python
temperatures=np.array([[22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],[19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]])
```

```python
print(temperatures)
```

```python
print("Shape:",temperatures.shape)
print("Data Type:",temperatures.dtype)
print("Number of Elements:",temperatures.size)
```

```python
week1=temperatures[0]
week2=temperatures[1]
weekends=temperatures[:,-2:]

print("Week 1:",week1)
print("Week 2;",week2)
print("Weekend temperatures for both weeks:",weekends)
```

# Pandas Series

```python
import pandas as pd
```

```python
marks=pd.Series([95, 92, 89, 85, 80],
                index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'])
print(marks)
```

```python
print(marks.iloc[0])
```

```python
print(marks.loc[['Rank1', 'Rank2', 'Rank3']])
```

```python
print(marks.iloc[2])
```

```python
print(marks[marks>90])
```

```python
marks.loc['Rank1']=100
print(marks)
```

```python
marks=marks.drop('Rank5')
print(marks)
```

```python
cgpa=marks/10
print(cgpa)
```

# Pandas DataFrame

```python
transactions=pd.DataFrame({"TransactionID":[101,102,103,104,105,106,107,108,109,110] ,
                           "ProductCategory":['Electronics','Clothing','Electronics','Furniture','Clothing','Electronics','Furniture','Clothing','Furniture', 'Electronics'],
                           "Region":['North','South','North','East','West','North','East','West','South','North'], 
                           "Amount":[200,150,300,450,200,250,300,180,350,400]})
print(transactions)
```

```python
print(transactions)
```

```python
print(transactions.head())
```

```python
print(transactions.tail())
```

```python
print(transactions.shape)
```

```python
print(transactions.columns)
```

```python
print(transactions.dtypes)
```

```python
print(transactions[["ProductCategory","Amount"]])
```

```python
print(transactions.iloc[:,-3:])
```

```python
print(transactions.loc[:,["ProductCategory","Region","Amount"]])
```

```python
filtered_data=transactions[(transactions["Region"]=="North") & (transactions["Amount"]>200)]
print(filtered_data)
```

```python
print(transactions["ProductCategory"].value_counts())
```

```python
print(transactions["Region"].unique())
```

```python
region_mean=transactions.groupby("Region")["Amount"].mean()
print(region_mean)
```

```python
transactions.loc[transactions["TransactionID"]==102,"Amount"]=165
print(transactions)
```

```python
transactions["Discount"]=transactions["Amount"]*0.10
print(transactions)
```

```python
transactions=transactions[transactions["TransactionID"]!=109]
print(transactions)
```

```python
transactions=transactions.drop(columns=["Discount"])
print(transactions)
```
